```python
import pandas as pd
import numpy as np
from scipy.stats import norm
users = pd.read_csv('marketplace_synthetic.csv')
funnel = users.groupby('device').agg(users=('user_id','count'), views=('viewed_listing','sum'), contacts=('contacted_seller','sum'))
funnel['view_to_contact_pct'] = 100 * funnel.contacts / funnel.views
print(funnel)
ab = pd.read_csv('ab_synthetic.csv')
s = ab.groupby('group').contacted_seller.agg(['sum','count','mean'])
p0, p1 = s.loc['control','mean'], s.loc['treatment','mean']
n0, n1 = s.loc['control','count'], s.loc['treatment','count']
pooled = s['sum'].sum()/s['count'].sum()
z = (p1-p0)/np.sqrt(pooled*(1-pooled)*(1/n0+1/n1))
pvalue = 2*norm.sf(abs(z))
se = np.sqrt(p0*(1-p0)/n0+p1*(1-p1)/n1)
print(s)
print('Difference (pp):',100*(p1-p0),'p-value:',pvalue)
print('95% CI (pp):',100*np.array([p1-p0-1.96*se,p1-p0+1.96*se]))
```
